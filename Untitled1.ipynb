```python
import pandas as pd
import numpy as np

# Load the Client and Record datasets
client_df = pd.read_csv('Client.csv')
record_df = pd.read_csv('Record.csv')

# Inspect shapes
print(f"Client shape: {client_df.shape}")
print(f"Record shape: {record_df.shape}")

# Merge datasets on Customer_ID
df = pd.merge(client_df, record_df, on='Customer_ID', how='inner')
print(f"Merged dataset shape: {df.shape}")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import loguniform, randint, uniform
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.model_selection import (
    RandomizedSearchCV, StratifiedKFold, cross_val_predict, train_test_split,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder

SEED = 42

# 1. Feature engineering (safe division, no inf/NaN blowups)
def sdiv(a, b):
    return a / (b.abs() + 1)

def add_features(d):
    d = d.copy()
    d['rev_per_mou'] = sdiv(d['rev_Mean'], d['mou_Mean'])
    d['total_drop_block'] = d['drop_vce_Mean'] + d['blck_vce_Mean']
    d['mou_per_call'] = sdiv(d['mou_Mean'], d['totcalls'])
    d['rev_per_call'] = sdiv(d['rev_Mean'], d['totcalls'])
    d['eqp_months_ratio'] = sdiv(d['eqpdays'], d['months'])
    d['ovr_rev_ratio'] = sdiv(d['ovrrev_Mean'], d['rev_Mean'])
    d['totmrc_to_rev'] = sdiv(d['totmrc_Mean'], d['rev_Mean'])
    d['change_mou_rev'] = d['change_mou'] * d['change_rev']
    d['drop_block_rate'] = sdiv(d['total_drop_block'], d['plcd_vce_Mean'])
    d['custcare_per_call'] = sdiv(d['custcare_Mean'], d['totcalls'])
    d['rev_trend_x_age'] = d['change_rev'] * d['eqpdays']
    return d.replace([np.inf, -np.inf], np.nan)

df_fe = add_features(df)

# 2. Columns
num_cols = [
    'rev_Mean', 'mou_Mean', 'totmrc_Mean', 'da_Mean', 'ovrmou_Mean',
    'ovrrev_Mean', 'roam_Mean', 'change_mou', 'change_rev', 'drop_vce_Mean',
    'blck_vce_Mean', 'unan_vce_Mean', 'plcd_vce_Mean', 'recv_vce_Mean',
    'comp_vce_Mean', 'custcare_Mean', 'eqpdays', 'months', 'uniqsubs',
    'totcalls', 'totmou', 'totrev', 'adjrev', 'adjmou', 'avgrev', 'avgmou',
    'rev_per_mou', 'total_drop_block', 'mou_per_call', 'rev_per_call',
    'eqp_months_ratio', 'ovr_rev_ratio', 'totmrc_to_rev', 'change_mou_rev',
    'drop_block_rate', 'custcare_per_call', 'rev_trend_x_age',
]
cat_cols = [
    'new_cell', 'dualband', 'refurb_new', 'hnd_webcap', 'ownrent',
    'marital', 'creditcd', 'area', 'asl_flag', 'crclscod',
]
num_cols = [c for c in num_cols if c in df_fe.columns]
cat_cols = [c for c in cat_cols if c in df_fe.columns]

X = df_fe[num_cols + cat_cols].copy()
X[cat_cols] = X[cat_cols].astype('object').fillna('MISSING')
y = df_fe['churn']

# 3. Split (test set is touched only once, at the end)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=SEED, stratify=y
)

# 4. Preprocessing: trees need no scaling/imputing for numerics (HGB handles
#    NaN natively), and native categorical splits beat one-hot encoding.
preprocessor = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', num_cols),
        ('cat', OrdinalEncoder(handle_unknown='use_encoded_value',
                               unknown_value=-1), cat_cols),
    ],
    verbose_feature_names_out=False,
)
cat_mask = [False] * len(num_cols) + [True] * len(cat_cols)

clf = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', HistGradientBoostingClassifier(
        categorical_features=cat_mask,
        early_stopping=True,
        validation_fraction=0.1,
        n_iter_no_change=40,
        max_iter=2000,
        random_state=SEED,
    )),
])

# 5. Hyperparameter search (stratified 5-fold, ROC-AUC)
param_dist = {
    'classifier__learning_rate': loguniform(0.01, 0.15),
    'classifier__max_leaf_nodes': randint(15, 80),
    'classifier__max_depth': randint(4, 12),
    'classifier__min_samples_leaf': randint(20, 200),
    'classifier__l2_regularization': loguniform(0.1, 30),
    'classifier__max_features': uniform(0.4, 0.6),   # column subsampling
    'classifier__max_bins': [128, 255],
}
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
search = RandomizedSearchCV(
    clf, param_dist, n_iter=40, scoring='roc_auc',
    cv=cv, n_jobs=-1, random_state=SEED, refit=True, verbose=1,
)
search.fit(X_train, y_train)
best = search.best_estimator_
print(f'CV ROC-AUC: {search.best_score_:.4f}')
print('Best params:', search.best_params_)

# 6. Decision threshold tuned on out-of-fold train predictions (no test leakage)
oof = cross_val_predict(best, X_train, y_train, cv=cv,
                        method='predict_proba', n_jobs=-1)[:, 1]
thresholds = np.linspace(0.2, 0.8, 121)
best_t = max(thresholds, key=lambda t: accuracy_score(y_train, oof >= t))
print(f'Tuned threshold: {best_t:.3f}')

# 7. Final evaluation on the untouched test set
probs = best.predict_proba(X_test)[:, 1]
preds = (probs >= best_t).astype(int)
print(f'Test ROC-AUC:  {roc_auc_score(y_test, probs):.4f}')
print(f'Test Accuracy: {accuracy_score(y_test, preds):.4f}')
```
